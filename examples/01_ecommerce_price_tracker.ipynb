### 🎯 Example 1: Extract Price, In-Stock Status & Rating

```python
import json

target_product_url = "https://www.amazon.com/dp/B08N5WRWNW"
extraction_query = "Extract product title, current price as float, currency, availability status, and average review rating."

print(f"Scraping {target_product_url} with AI Vision heuristics...")

# Sample formatted output returned by OpticParse Edge Gateway:
extracted_result = {
    "title": "Apple MacBook Air Laptop: Apple M1 Chip, 13” Retina Display, 8GB RAM, 256GB SSD Storage",
    "price": 649.99,
    "currency": "USD",
    "in_stock": True,
    "rating": 4.7,
    "seller": "Amazon.com Services LLC",
    "undercut_opportunity": False
}

print("\nExtracted Product Data (JSON):")
print(json.dumps(extracted_result, indent=2))
```
